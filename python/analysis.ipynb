# Food Delivery Data Analysis
Analyze order volume, cancellations, delivery performance, ratings and order value.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/food_delivery_orders.csv', parse_dates=['Order_Date'])
df.head()
```

```python
print(df.shape)
print(df.isna().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
kpis = {'Total Orders': df.Order_ID.nunique(), 'Net Order Value': df.Net_Value.sum(), 'Average Delivery Time': df.Delivery_Time_Min.mean(), 'Average Rating': df.Customer_Rating.mean(), 'Cancellation Rate %': df.Cancelled_Flag.mean()*100}
pd.Series(kpis)
```

```python
city = df.groupby('City').agg(Orders=('Order_ID','count'), Net_Value=('Net_Value','sum'), Avg_Delivery=('Delivery_Time_Min','mean'), Avg_Rating=('Customer_Rating','mean'), Cancellation_Rate=('Cancelled_Flag','mean')).sort_values('Orders', ascending=False)
city['Cancellation_Rate'] *= 100
city
```

```python
cuisine = df.groupby('Cuisine').agg(Orders=('Order_ID','count'), Net_Value=('Net_Value','sum'), Avg_Rating=('Customer_Rating','mean')).sort_values('Net_Value', ascending=False)
cuisine
```

```python
df['Delivery_Band'] = pd.cut(df.Delivery_Time_Min, [0,30,45,60,100], labels=['0-30 min','31-45 min','46-60 min','60+ min'])
band = df.groupby('Delivery_Band', observed=True).agg(Orders=('Order_ID','count'), Avg_Rating=('Customer_Rating','mean'))
band
```

```python
monthly = df.set_index('Order_Date').resample('MS').agg(Orders=('Order_ID','count'), Net_Value=('Net_Value','sum'))
monthly
```
